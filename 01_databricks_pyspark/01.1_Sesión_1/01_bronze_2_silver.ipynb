```python
spark.sql("create volume if not exists workspace.default.silver")
```

```python
clientes = spark.read.\
    option("header",True).\
        option("inferSchema",True).csv("/Volumes/workspace/default/bronze/clientes.csv")
```

```python
compras = spark.read.option("header",True).option("inferSchema",True).csv("/Volumes/workspace/default/bronze/compras.csv")
```

```python
clientes_silver = clientes.filter(clientes.pais=="España")
```

```python
compras_silver = compras.filter(compras.pais=="España")
```

```python
clientes_silver.write.mode("overwrite").format("avro").save("/Volumes/workspace/default/silver/dim_clientes")
```

```python
compras_silver.write.mode("overwrite").format("avro").save("/Volumes/workspace/default/silver/fct_compras")
```
